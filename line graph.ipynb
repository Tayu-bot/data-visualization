```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [10,20,30,40,50,60]
plt.plot(x,y)
plt.show()
```

```python
# multiple line in one graph
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y1=[10,20,30,40,50,60]
y2=[60,50,40,30,20,10]
plt.plot(x,y1, label='line 1')
plt.plot(x,y2, label='line 2')
plt.title('multiple lines example')
plt.xlabel('x-axis')
plt.ylabel('Y-axis')
plt.legend()    #show legend
plt.grid(True)
plt.show()
```

```python
#line color
import matplotlib.pyplot as plt
x=(1,2,3,4,5,6)
y=(10,20,30,40,50,60)
plt.plot(x,y, color = 'black', label = 'green line')
plt.plot(x,y, color='blue', linestyle='--', label='blue dashed line')
plt.title('line color example')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# makers in line plot
import matplotlib.pyplot as plt
x=(1,2,3,4,5,6)
y=(10,20,30,40,50,60)
plt.plot(x,y, marker = 'o', color = 'b',
    markersize = 8, markeredgecolor = 'r',
    markerfacecolor ='yellow', linewidth=2)
plt.title('line plot with markers')
plt.xlabel('x-axis')
plt.ylabel('y-label')
plt.grid(True)
plt.show()
```

```python
#line style
import matplotlib.pyplot as plt
x=[1,2,3,4,5,]
y=[10,20,30,40,50]
plt.plot(x,y, linestyle ='-', label='solid line')
plt.plot(x,y, linestyle ='--', label='Dashed line')
plt.plot(x,y, linestyle =':', label='dotted line')
plt.plot(x,y, linestyle ='-.', label='Dash-dot line')
plt.title('line styles example')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Bar chart(vertical ) plt.bar()
import matplotlib.pyplot as plt
```

```python
categories = ['jan', 'feb', 'mar', 'apri', 'may']
Sales = [20, 35, 30, 45, 40]
plt.bar(categories, Sales, color = 'skyblue',
       edgecolor = 'black', width = 0.6)
plt.title('monthly sales bar chart')
plt.xlabel('month')
plt.ylabel('Sales(in 1000s)')
plt.grid(axis = 'y', linestyle = '--', alpha = 0.7)
plt.show()
```
