```python
import matplotlib.pyplot as plt
```

```python
products = ['phone', 'laptop','smartwatch', 'tablet']
sales = ['550', '400', '350', '450']
```

```python
plt.barh(products, sales, color ='lightgreen', edgecolor='black')
plt.title('product sales (horizontal bar chart)')
plt.xlabel('sales (in units')
plt.ylabel(products)
plt.grid(axis='x', linestyle='--',alpha=0.7)
plt.show()
```
